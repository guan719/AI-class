```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report
```

```python
X = df.drop("Wine", axis=1)
y = df["Wine"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

Regression

```python
X_reg = df.drop(["Alcohol", "Wine"], axis=1)
y_reg = df["Alcohol"]

X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg,
    y_reg,
    test_size=0.2,
    random_state=0
)

X_train_reg.shape, X_test_reg.shape, y_train_reg.shape, y_test_reg.shape
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

model_lr_reg = LinearRegression()

model_lr_reg.fit(X_train_reg, y_train_reg)

pred_lr_reg = model_lr_reg.predict(X_test_reg)

print("MSE:", mean_squared_error(y_test_reg, pred_lr_reg))
print("R2:", r2_score(y_test_reg, pred_lr_reg))
```

```python
from sklearn.tree import DecisionTreeRegressor

model_dt_reg = DecisionTreeRegressor(random_state=0)

model_dt_reg.fit(X_train_reg, y_train_reg)

pred_dt_reg = model_dt_reg.predict(X_test_reg)

print("MSE:", mean_squared_error(y_test_reg, pred_dt_reg))
print("R2:", r2_score(y_test_reg, pred_dt_reg))
```

```python
from sklearn.ensemble import RandomForestRegressor

model_rf_reg = RandomForestRegressor(random_state=0)

model_rf_reg.fit(X_train_reg, y_train_reg)

pred_rf_reg = model_rf_reg.predict(X_test_reg)

print("MSE:", mean_squared_error(y_test_reg, pred_rf_reg))
print("R2:", r2_score(y_test_reg, pred_rf_reg))
```

```python
print("Linear Regression")
print("MSE:", mean_squared_error(y_test_reg, pred_lr_reg))
print("R2 :", r2_score(y_test_reg, pred_lr_reg))

print("\nDecision Tree Regression")
print("MSE:", mean_squared_error(y_test_reg, pred_dt_reg))
print("R2 :", r2_score(y_test_reg, pred_dt_reg))

print("\nRandom Forest Regression")
print("MSE:", mean_squared_error(y_test_reg, pred_rf_reg))
print("R2 :", r2_score(y_test_reg, pred_rf_reg))
```
