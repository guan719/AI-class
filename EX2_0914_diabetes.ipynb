Classifiation 4개

```python
X = df.drop("Outcome", axis=1)
y = df["Outcome"]
```

```python
X_train,X_test, y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=0)
```

```python
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
model_dt = DecisionTreeClassifier()
model_dt.fit(X_train, y_train)
```

```python
pred_dt = model_dt.predict(X_test)
print(classification_report(y_test, pred_dt))
```

```python
model_rf= RandomForestClassifier()
model_rf.fit(X_train, y_train)
```

```python
pred_rf = model_rf.predict(X_test)
print(classification_report(y_test, pred_rf))
```

```python
model_lr = LogisticRegression(max_iter=200)
model_lr.fit(X_train, y_train)
```

```python
pred_lr = model_lr.predict(X_test)
print(classification_report(y_test, pred_lr))
```

```python
model_kn = KNeighborsClassifier()
model_kn.fit(X_train, y_train)
```

```python
pred_kn = model_kn.predict(X_test)
print(classification_report(y_test, pred_kn))
```

Regression 3개

```python
X_reg = df.drop(["BMI", "Outcome"], axis=1)
y_reg = df["BMI"]
```

```python
from sklearn.model_selection import train_test_split

X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg,
    y_reg,
    test_size=0.2,
    random_state=0
)
```

```python
X_train_reg.shape, X_test_reg.shape, y_train_reg.shape, y_test_reg.shape
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

model_lr_reg = LinearRegression()

model_lr_reg.fit(X_train_reg, y_train_reg)

pred_lr_reg = model_lr_reg.predict(X_test_reg)

print("MSE:", mean_squared_error(y_test_reg, pred_lr_reg))
print("R2:", r2_score(y_test_reg, pred_lr_reg))
```

```python
from sklearn.tree import DecisionTreeRegressor

model_dt_reg = DecisionTreeRegressor()

model_dt_reg.fit(X_train_reg, y_train_reg)

pred_dt_reg = model_dt_reg.predict(X_test_reg)

print("MSE:", mean_squared_error(y_test_reg, pred_dt_reg))
print("R2:", r2_score(y_test_reg, pred_dt_reg))
```

```python
from sklearn.ensemble import RandomForestRegressor

model_rf_reg = RandomForestRegressor()

model_rf_reg.fit(X_train_reg, y_train_reg)

pred_rf_reg = model_rf_reg.predict(X_test_reg)

print("MSE:", mean_squared_error(y_test_reg, pred_rf_reg))
print("R2:", r2_score(y_test_reg, pred_rf_reg))
```

```python
print("Linear Regression")
print("MSE:", mean_squared_error(y_test_reg, pred_lr_reg))
print("R2 :", r2_score(y_test_reg, pred_lr_reg))

print("\nDecision Tree")
print("MSE:", mean_squared_error(y_test_reg, pred_dt_reg))
print("R2 :", r2_score(y_test_reg, pred_dt_reg))

print("\nRandom Forest")
print("MSE:", mean_squared_error(y_test_reg, pred_rf_reg))
print("R2 :", r2_score(y_test_reg, pred_rf_reg))
```
